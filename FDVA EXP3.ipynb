```python

# Import the pandas library
import pandas as pd

# Create a dataset containing missing values and duplicates
data = {
    'Name': ['Arun', 'Priya', 'Kumar', 'Divya', 'Priya'],
        'Age': [20, 21, None, 22, 21],
            'Marks': [85, 90, 78, None, 90]
            }
 # Create a DataFrame
df = pd.DataFrame(data)
# Display the original dataset
print("Original Dataset:")
print(df)
# Check for missing values
print("\nMissing Values:")
print(df.isnull().sum())
# Remove duplicate rows
df = df.drop_duplicates()
print("\nAfter Removing Duplicates:")
print(df)
 # Fill missing Age values with the mean
df['Age'] = df['Age'].fillna(df['Age'].mean())
# Fill missing Marks values with the mean
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
# Convert Age to integer
df['Age'] = df['Age'].round().astype(int)
# Remove extra spaces from names
df['Name'] = df['Name'].str.strip()
# Display the cleaned dataset
print("\nCleaned Dataset:")
print(df)
# Check missing values after cleaning
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
```
