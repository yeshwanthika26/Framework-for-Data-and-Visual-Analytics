```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
# Create a sample dataset
data = {
    'Name': ['Arun', 'Priya', 'Kumar', 'Divya', 'Ravi'],
        'Marks': [85, 90, 78, 95, 88],
            'Age': [20, 21, 22, 20, 23]
            }
# Create a DataFrame
df = pd.DataFrame(data)
 # Display the dataset
print("Student Dataset:")
print(df)
# Create a bar chart for student marks
plt.figure(figsize=(8, 5))
plt.bar(df['Name'], df['Marks'])
plt.title('Student Marks')
plt.xlabel('Student Name')
plt.ylabel('Marks')
plt.show()
# Create a line chart for student marks
plt.figure(figsize=(8, 5))
plt.plot(df['Name'], df['Marks'], marker='o')
plt.title('Student Marks - Line Chart')
plt.xlabel('Student Name')
plt.ylabel('Marks')
plt.grid(True)
plt.show()
# Create a histogram for age distribution
plt.figure(figsize=(8, 5))
plt.hist(df['Age'], bins=5, edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
# Create a scatter plot for age and marks
plt.figure(figsize=(8, 5))
plt.scatter(df['Age'], df['Marks'])
plt.title('Age vs Marks')
plt.xlabel('Age')
plt.ylabel('Marks')
plt.show()
# Create a pie chart for student marks
plt.figure(figsize=(7, 7))
plt.pie(df['Marks'], labels=df['Name'], autopct='%1.1f%%')
plt.title('Marks Distribution')
plt.show()
```
